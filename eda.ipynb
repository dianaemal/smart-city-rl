```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
url = "https://data.edmonton.ca/resource/q7ua-agfg.csv?$limit=1000000"
df = pd.read_csv(url)
df.head()
```

```python
df.shape
df.info()
df.describe().T
```

```python
#missing values:
df.isna().sum()
```

```python
# check for unique values:
df.nunique()
```

```python
# count of service by service category:
service_count = df['service_category'].value_counts()
plt.figure(figsize=(8, 6))
plt.barh(service_count.index, service_count, color='deeppink')
plt.ylim(0, 10)
plt.title('Count Plot of number of services per category')
plt.xlabel('Count')
plt.ylabel('service_category')
plt.show()
```

```python
desc_count = df['service_description'].value_counts()
plt.figure(figsize=(8, 6))
plt.barh(desc_count.index, desc_count, color='deeppink')
plt.ylim(0, 10)
plt.title('Count Plot of number of service descriptions')
plt.xlabel('Count')
plt.ylabel('service_description')
plt.show()
```

```python

# Combine year and month into a single column
df['year_month'] = df['year'].astype(str) + '-' + df['month_number'].astype(str).str.zfill(2)  # e.g., "2021-march"

# Aggregate counts by year_month
monthly_counts = df.groupby('year_month').size().reset_index(name='request_count')

# Sort by time
monthly_counts = monthly_counts.sort_values('year_month')

# Plot
plt.figure(figsize=(20,6))
sns.lineplot(data=monthly_counts, x='year_month', y='request_count', marker='o')
plt.xticks(rotation=45)
plt.title("Edmonton 311 Requests Over Time")
plt.xlabel("Month")
plt.ylabel("Number of Requests")
plt.tight_layout()
plt.show()
```

```python
# pie chart of 
df = df.dropna(subset=['request_status'])
count = df['request_status'].value_counts()
plt.pie(count, labels=count.index, autopct='%1.1f%%', startangle=90)
# Add a title
plt.title('Distribution of request status')
# Ensure the circle is drawn as a circle
plt.axis('equal')
plt.legend()
# Display the chart
plt.show()
```
