```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def get_monthly(df, year, local_area):
    new_df = df[(
        df["Local area"] == local_area
    )]
    new_df["date"] = pd.to_datetime(new_df["date"])
    new_df_year = new_df[new_df["date"].dt.year == year]
    new_df_year["month"] = new_df_year["date"].dt.month

    monthly = (
        new_df_year
        .groupby("month")["request_count"]
        .mean()
    )

    print(monthly)
```

```python
def get_week(df, year, local_area):
    new_df = df[(
        df["Local area"] == local_area
    )]
    new_df["date"] = pd.to_datetime(new_df["date"])
    new_df_year = new_df[new_df["date"].dt.year == year]
    new_df_year["days_of_week"] = new_df_year["date"].dt.day_name()

    weekly_pattern  = (
        new_df_year
        .groupby("days_of_week")["request_count"]
        .mean()
    )
    pd.set_option('display.max_rows', None)
    print(weekly_pattern)
```

```python
daily_demand = pd.read_csv("../data/daily_demand.csv")
get_monthly(daily_demand, 2024, "Grandview-Woodland")
get_week(daily_demand, 2024, "Grandview-Woodland")
```

```python
daily_demand = pd.read_csv("../data/daily_demand.csv")

```

```python
daily_demand = daily_demand.sort_values(
    ["Local area", "date"]
)
```

```python
daily_demand["lag_1"] = daily_demand.groupby("Local area")["request_count"].shift(1)
daily_demand.tail()
```

```python
daily_demand["lag_1"] = daily_demand.groupby("Local area")["request_count"].shift(1)
daily_demand.tail(20)
```

```python
daily_demand["lag_7"] = daily_demand.groupby("Local area")["request_count"].shift(7)
daily_demand.tail(20)
```

```python
daily_demand["rolling_mean_7"] = daily_demand.groupby("Local area")["request_count"].transform(lambda x: x.rolling(7).mean())
daily_demand.tail()
```

```python
daily_demand["date"] = pd.to_datetime(
    daily_demand["date"],
    utc=False
)
```

```python
daily_demand["day_of_week"] = daily_demand["date"].dt.dayofweek
daily_demand.tail()
```

```python
daily_demand["month"] = daily_demand["date"].dt.month
daily_demand.tail()
```

```python
daily_demand["weekend"] = (daily_demand["day_of_week"] >= 5).astype(int)
daily_demand.tail()
```

```python
daily_demand = daily_demand.dropna()
daily_demand.head()
```

```python
daily_demand["rolling_mean_7"] = daily_demand["rolling_mean_7"].astype(int)
daily_demand["lag_1"] = daily_demand["lag_1"].astype(int)
daily_demand["lag_7"] = daily_demand["lag_7"].astype(int)
```

```python
# save the new dataset
daily_demand.to_csv(
    "../data/forecasting_dataset.csv",
    index=False
)
```
